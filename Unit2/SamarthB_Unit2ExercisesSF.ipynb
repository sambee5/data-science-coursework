What is a prior? How are priors related to
- context?
- assumptions?
- predictions?

Priors are any sort of previously known info that could be helpful to a prediction. They are very similar to context, as context provides a basis for assumptions and predictions, just as priors do. - predictions?

A likelihood is a measure of how likely data is under certain assumptions. A likelihood without any outside information (like priors!) will make little to no sense because it seems irrelevant. Likelihoods are independent of context and assumptions, meaning they are just raw predictions that don't have much value on it's own.

**Task6**:

What is a posterior? How are posteriors related to:

- context?
- assumptions?
- predictions?

A posterior is a combination of likelihoods and priors to result in a prediction that is much more accurate, and uses pieces from both to qualify the chance of an event taking place. Mathematically speaking, posteriors can be simplified to priors multiplied by the likelihood. After lots of messy algebraic manipulation, we end up with an elegant equation that allows us to predict effectively and with confidence. Context and assumptions are two of the parameters used in posteriors, and allow us to create a prediction that is more well suited for a variety of situations.
