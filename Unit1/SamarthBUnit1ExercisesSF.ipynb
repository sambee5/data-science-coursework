**Task1**:

Describe the similarities and differences between an int and a float in python.

**Answer1**:

There are many similarities and differences between an int and a float in python. int, as the name suggests, is only able to hold the integer data type. For example, an int would be 2, 3, -1, or 0. You are able to iterate over a range of int's in a loop. Similarly, a float is a data type that holds numbers, but specifically decimal values. A principal characteristic of floats are that they have a decimal point. This means that 2.0, -5.6, and 9.2 are all floats. An int is represented by a max of 32 bits, while floats represent 64 bits. The int data type is more precise than floats are, which is why they are typically used for more fine operations.


**Task2**:

Describe the similarities and differences between: strings, lists, and dicts.

**Answer2**:

Strings, lists, and dicts are all built in data types in python. A string, which is common to other languages as well, is an immutable data type that is able to process character values like a, b, c and some other special characters. They are defined using double or single quotes. Lists are also common to other programming languages, but usually go by a different name: arrays. Lists contain many different data types, and each entry or element are seperated by commas. The entire list is enclosed by square brackets. Lists are mutable, which mean they can be modified after being first defined, and elements can be added, rearranged, or removed. Strings and lists intersect, as you could return the index (spot an element is in) and and it would return that specific character as if it was a list. Strings and lists can both be spliced, using the colon and specifying the range of elements or characters desired. Finally, dicts or dictionaries are also common to other programming languages and typically go under the name of hashmap. In essence, dicts are very similar to lists, in that they have different elements but they also require an additional piece of info to access those elements: a key. Dictionaries are defined with a key and a value. To access a value, a key must be provided. Like lists, dictionaries are also mutable. Dictionaries are bounded by curly braces. Each of these 3 data types have their own in-built commands that are included with python.

**Task3**:

What is a numpy array? Why would we ever use one over a list or dict?

**Answer3**:

There is extensive support towards the python libraries as they are commonly open source. This gives them lots of room for development, making them more useful than simple in built data structures like lists or dictionaries. A numpy array has carries forward many of the same capabilities as their in-built counterpart. However they have some very useful vectorized operations that make them more mathematically efficient to use than lists or dictionaries. A numpy array operates more like a matrix in math, rather than just sticking to the standard programming definition. When performing operations on numpy arrays, every element is converted to their own datatype, allowing operations to be done faster. Lists and dictionaries can technically do this, but it's slow, and inferior to what numpy is capable of doing. It's efficiency allows it to be very scalale and applicable to more realistic datasets.

**Task4**:

What is a pandas dataframe? Why would we ever want to use one over a numpy array?

**Answer4**:

Pandas is a library that is built on top of numpy, meaning it carries forward similar vectorized operations. A pandas dataframe organizes data as if its a able, including labeled rows and columns. This makes it a useful tool when processing large datasets like those written in csv and json. One can navigate across a dataframe using labeled rows and columns, unlike indexes like python. Pandas is easily compatible with already existing datasets because they have their own methods like read_csv which make it easy to wrangle data in comparison to numpy arrays.

**Task5**:

What are for loops, and why would we ever use them?

**Answer5**:

For loops are a tool that repeat an action a specified amount of times or until a condition met. They're very useful if an action needs to be repeated multiple times. Using for loops saves memory in the long run because it would take less lines to perform the same task. For loops can be applied to preexisting data structures like lists, and also custom entities like numpy arrays. When running a for loop through a list, you are "iterating" through that list. You're accessing each index of that list for every time the code loops. Pandas has a better alternative to a for loop, the query() method, but the traditional for loop is still very applicable in other libraries like numpy.
